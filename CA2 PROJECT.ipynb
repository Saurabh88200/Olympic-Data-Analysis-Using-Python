```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Barchart

plt.figure(figsize=(10,5))

plt.hist(df['Age'].dropna(), bins=20)

plt.title("Age Distribution of Olympic Athletes")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
#Histogram

plt.figure(figsize=(10,5))

plt.hist(df['Age'].dropna(), bins=20)

plt.title("Age Distribution of Olympic Athletes")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
#LineChart
year_count = df.groupby('Year').size()

plt.figure(figsize=(12,5))

plt.plot(year_count.index, year_count.values, marker='o')

plt.title("Olympic Participation Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Participants")

plt.grid()
plt.show()
```

```python
#ScatterPlot
plt.figure(figsize=(8,5))

plt.scatter(df['Age'], df['Year'])

plt.title("Age vs Olympic Year")
plt.xlabel("Age")
plt.ylabel("Year")

plt.show()
```

```python
#pieChart
medal_counts = df['Medal'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    medal_counts.values,
    labels=medal_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Olympic Medal Percentage")

plt.show()

```

```python
#ScatterPlot
#CountPlot
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='Medal')

plt.title("Medal Count")
plt.show()
```

```python
#Box Plot
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x='Medal', y='Age')

plt.title("Age Distribution by Medal")
plt.show()
```

```python
#HeatMap
plt.figure(figsize=(12,8))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
#PairPlot
sns.pairplot(
    df.select_dtypes(include=np.number).sample(
        min(300, len(df)),
        random_state=42
    )
)

plt.show()
```
